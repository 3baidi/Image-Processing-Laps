```python
import os
import cv2
from skimage import data

os.makedirs('images', exist_ok=True)


cameraman_img = data.camera()
cv2.imwrite('images/cameraman.tif', cameraman_img)


moon_img = data.moon()  
cv2.imwrite('images/lena_gray_256.tif', moon_img)

```

```python
import cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from PIL import Image
import numpy as np


img = cv2.imread('images/cameraman.tif')
plt.figure()
plt.imshow(img, cmap='gray')
plt.title("camera")
plt.show()


img2 = Image.open('images/lena_gray_256.tif')
plt.figure()
plt.imshow(np.array(img2), cmap=cm.Greys_r)
plt.title("PIL Image")
plt.show()
```

```python

cv2.imwrite('new_image.jpg', img)


img2.save('new_image2.jpg')
```
